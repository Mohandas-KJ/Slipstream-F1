## 03: Tyre and Stint Strategy Analysis

```python
# Setup & Imports
import sys
from pathlib import Path
sys.path.append("../..")

import Sliplib.slipload as sl
import Sliplib.selection as slipclassify
import Sliplib.GridSum as grid_sum
import Sliplib.plotting as slplot
import Sliplib.Gotyres as GT
from configuration.writer import load_battles

# Load race-specific contenders and rival pairs
DATA = load_battles("../data/battles.json")
```

```python
# Load Race Session
YEAR = 2025
GP_NAME = "Austrian Grand Prix"  # Change to target Grand Prix name

session = sl.load_gp(YEAR, GP_NAME, "R")
session.load()
r_laps = session.laps
```

### 1. Grid-Wide Stint Overview
Breakdown of total stints, compounds, and pit stop counts across the entire grid.

```python
# Calculate Stints and Pit Stops for all Drivers
grid_stint = grid_sum.get_stint(r_laps)
grid_stint.to_csv("../data/Race_Stint_Summary.csv", index=False)
grid_stint
```

### Insights:
- *(Record 1-stop vs 2-stop trends, tyre compound choices, outlier strategies)*

### 2. Contender Stint & Compound Breakdown

```python
# Filter Contenders & Stint Data
top_drivers = slipclassify.select_drivers(r_laps, DATA.CONTENDERS)
drivers_tyre_data = slipclassify.pick_tyre_data(top_drivers)

# Display compound, lap range, and freshness breakdown
GT.stint_breakdown(drivers_tyre_data)
```

### 3. Tyre Degradation Analysis
Plot quicklap pace vs tyre life to evaluate degradation slope.

```python
# Tyre Degradation Comparison for Rival Pairs
for d1, d2 in DATA.RIVALS:
    slplot.plot_tyre_degradation(
        top_drivers,
        (d1, d2),
        save_path=f"../plots/{d1}_{d2}_tyre_deg.png"
    )
```
