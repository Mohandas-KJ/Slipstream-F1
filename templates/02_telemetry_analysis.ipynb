## 02: Telemetry Analysis

```python
# Setup & Imports
import sys
from pathlib import Path
sys.path.append("../..")

import Sliplib.slipload as sl
import Sliplib.selection as slipclassify
import Sliplib.plotting as slplot
from configuration.writer import load_battles

# Load race-specific contenders and rival pairs
DATA = load_battles("../data/battles.json")
print("Loaded Contenders:", DATA.CONTENDERS)
print("Loaded Rivals:", DATA.RIVALS)
```

```python
# Load Race Session
YEAR = 2025
GP_NAME = "Austrian Grand Prix"  # Change to target Grand Prix name

session = sl.load_gp(YEAR, GP_NAME, "R")
session.load()
r_laps = session.laps
```

### 1. Fastest Lap Selection & Summary

```python
# Extract Lap Data for Contenders
top_drivers = slipclassify.select_drivers(r_laps, DATA.CONTENDERS)

# Pick Fastest Lap for each contender
fastest_data = slipclassify.pick_fastest_group(top_drivers)

# Fast Lap Summary
fast_sum = slipclassify.tel_fast_lap_summary(fastest_data.values())
fast_sum
```

```python
# Extract Car Telemetry with Distance Channel
tel_dist_datas = slipclassify.tel_get_for_all(fastest_data)
```

### 2. Head-to-Head Telemetry Comparisons (Distance vs Speed / Brake / Gear / RPM)

```python
# Distance vs Speed Comparison for All Rival Pairs
for d1, d2 in DATA.RIVALS:
    slplot.plot_dist_speed(
        [tel_dist_datas[d1], tel_dist_datas[d2]],
        [d1, d2],
        save_path=f"../plots/{d1}_{d2}_speed.png"
    )
```

```python
# Distance vs Brake Application
for d1, d2 in DATA.RIVALS:
    slplot.plot_dist_brake(
        [tel_dist_datas[d1], tel_dist_datas[d2]],
        [d1, d2],
        save_path=f"../plots/{d1}_{d2}_brake.png"
    )
```

```python
# Distance vs Gear Selection
for d1, d2 in DATA.RIVALS:
    slplot.plot_dist_ngear(
        [tel_dist_datas[d1], tel_dist_datas[d2]],
        [d1, d2],
        save_path=f"../plots/{d1}_{d2}_gear.png"
    )
```

```python
# Distance vs Engine RPM
for d1, d2 in DATA.RIVALS:
    slplot.plot_dist_rpms(
        [tel_dist_datas[d1], tel_dist_datas[d2]],
        [d1, d2],
        save_path=f"../plots/{d1}_{d2}_rpm.png"
    )
```

### 3. Optional: Lap Telemetry Dashboard
Generate full 3-panel dashboard (Speed, Gear, Throttle/Brake) for standout drivers.

```python
import Sliplib.dashboard as sldash

# Example: Generate dashboard for the fastest driver
fastest_driver = fast_sum.sort_values(by="LapTime").iloc[0]["Driver"]
sldash.create_dashboard(
    tel_dist_datas[fastest_driver],
    f"{fastest_driver} - Fastest Lap",
    save_path=f"../plots/{fastest_driver}_dashboard.png"
)
```
